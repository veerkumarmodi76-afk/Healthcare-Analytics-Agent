```python
"""
inspect_portfolio_outputs.py

Run this from the notebooks folder.

Prints for each Portfolio Analytics output:
1. Shape
2. Columns
3. First 20 rows

Finally prints a master summary table.
"""

from pathlib import Path
import pandas as pd

# ------------------------------------------------------------------
# Project Paths
# ------------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent
PORTFOLIO_DIR = PROJECT_ROOT / "outputs" / "portfolio"

print("=" * 120)
print("Portfolio Directory:")
print(PORTFOLIO_DIR)
print("Directory Exists:", PORTFOLIO_DIR.exists())
print("=" * 120)

# ------------------------------------------------------------------
# Files to Inspect
# ------------------------------------------------------------------

FILES = [
    "executive_kpis.csv",
    "portfolio_metrics.csv",
    "portfolio_health.csv",

    "risk_distribution.csv",
    "risk_by_age_band.csv",
    "risk_by_channel.csv",
    "risk_by_policy_type.csv",
    "risk_profitability.csv",
    "risk_claims_summary.csv",
    "risk_premium_summary.csv",

    "frequency_by_age.csv",
    "severity_by_age.csv",
    "top_loss_segments.csv",

    "premium_trend.csv",
    "claims_trend.csv",
    "loss_ratio_trend.csv",
    "lapse_trend.csv",
    "portfolio_growth_summary.csv",
    "risk_mix_trend.csv",

    "retention_summary.csv",
]

# ------------------------------------------------------------------
# Inspection
# ------------------------------------------------------------------

master_summary = []

for file in FILES:

    file_path = PORTFOLIO_DIR / file

    print("\n" + "=" * 120)
    print(file)
    print("=" * 120)

    if not file_path.exists():
        print("❌ File not found:", file_path)
        continue

    df = pd.read_csv(file_path)

    print(f"\nShape: {df.shape}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 20 Rows:")
    print(df.head(20))

    master_summary.append({
        "File": file,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Column Names": ", ".join(df.columns.tolist())
    })

# ------------------------------------------------------------------
# Master Summary
# ------------------------------------------------------------------

print("\n" + "=" * 120)
print("MASTER SUMMARY")
print("=" * 120)

master_df = pd.DataFrame(master_summary)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)

print(master_df)
```
