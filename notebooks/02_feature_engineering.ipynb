```python
import pandas as pd
df = pd.read_csv("../data/processed/processed_data.csv")
```

```python
missing_df = (
    df.isna()
      .sum()
      .reset_index()
      .rename(columns={'index': 'column_name', 0: 'missing_count'})
)

missing_df = missing_df[missing_df['missing_count'] > 0]
print(missing_df)
```
